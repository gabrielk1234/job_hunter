## 寫入職缺的詳細資訊

```python
file_name = 'all_jobs_detail.json'

# 1. 先把舊資料讀出來（如果有這個檔案的話）
if os.path.exists(file_name):
    with open(file_name, 'r', encoding='utf-8') as f:
        all_jobs_detail_dict = json.load(f)
else:
    # 第一次跑，連檔案都還沒有，就建一個空的字典
    all_jobs_detail_dict = {} 

# ----------------- 爬蟲抓資料的迴圈開始 -----------------
for job in response.json()['data']:
    job_id = job['jobNo']
    job_code = job['link']['job'].split('/')[-1]
    # 去request職缺的細節
    job_detail_response = requests.get(f'https://www.104.com.tw/api/jobs/{job_code}', cookies=cookies, headers=headers)
    
    if job_detail_response.status_code == 200:
        job_detail_json = job_detail_response.json().get('data')
    else: 
        print(f'{job_code}獲取資料失敗')
        continue

    # 2. 開始判斷是不是重複的
    if job_id in all_jobs_detail_dict:
        print(f"👀 職缺 {job_id} 已經存過囉，更新一下資料！")
        # 直接覆蓋它！這樣可以順便更新「最新應徵人數(applyCnt)」這類會變動的數字
        all_jobs_detail_dict[job_id] = job_detail_json 
    else:
        print(f"✨ 發現新職缺 {job_id}！加進去！")
        all_jobs_detail_dict[job_id] = job_detail_json
        
    # 因為有request，要適當休息一下 隨機2到5秒
    time.sleep(rd.randint(2,5))
# ----------------- 爬蟲抓資料的迴圈結束 -----------------

# 3. 把更新後的超大字典，重新存回檔案裡
with open(file_name, 'w', encoding='utf-8') as f:
    json.dump(all_jobs_detail_dict, f, ensure_ascii=False, indent=4)
```

```python
import re
def create_chunks(job_id, full_document, base_metadata):
    
    # 2. 準備用來戴帽子的全局變數
    cust_name = base_metadata['custName']
    job_name = base_metadata['jobName']
    
    # 3. 用正規表達式 (Regex) 按 "## " 把文章切開
    # 這邊會切出包含標題和內容的 list
    raw_sections = re.split(r'\n## ', full_document)
    
    chunks_for_chroma = []
    
    for idx,section in enumerate(raw_sections):
        # 去除掉可能多餘的空白
        section = section.strip()
        if not section or section.startswith('#'):
            # 跳過最前面的 "# 允和科技股份有限公司 - Unity資深研發工程師" 這個大標題
            continue
            
        # 把標題跟內容分開 (例如 "公司資訊\n公司名稱:...")
        lines = section.split('\n', 1)
        if len(lines) < 2:
            continue
            
        section_title = lines[0].strip()
        section_content = lines[1].strip()
        
        # 4. 戴帽子！組合全局資訊
        enriched_text = f"【{cust_name} - {job_name}】的{section_title}：\n{section_content}"
        
        # 5. 準備這一個 chunk 的專屬 metadata
        chunk_meta = base_metadata.copy()
        chunk_meta['job_id'] = job_id
        chunk_meta['chunk_type'] = section_title # 順便記一下這塊是福利還是條件
        
        ids = f"{job_id}_{idx}"
        
        chunks_for_chroma.append({
            "text": enriched_text,
            "metadata": chunk_meta,
            "ids":ids
        })
        
    return chunks_for_chroma
```
